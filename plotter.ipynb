```python
import matplotlib.pyplot as plt

# Enable dark background styling
plt.style.use('dark_background')

def generate_chart(data_dict, filename):
    # CHANGED: Increased width, decreased height
    fig, ax = plt.subplots(figsize=(5, 3.5)) 
    
    categories = list(data_dict.keys())
    values = list(data_dict.values())
    
    # Colors: Red for 'No', Blue for 'Yes'
    colors = ['#ff4d4d', '#4da6ff']
    
    # Create the bars
    bars = ax.bar(categories, values, color=colors, width=0.6)
    
    # Formatting to keep it extremely clean (no captions, minimal borders)
    ax.set_ylim(0, 1.05)
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)
    ax.spines['left'].set_color('#444444')
    ax.spines['bottom'].set_color('#444444')
    
    # Make tick labels slightly larger for the presentation
    ax.tick_params(axis='x', colors='white', labelsize=14)
    ax.tick_params(axis='y', colors='white', labelsize=12)
    
    ax.set_title('')
    ax.set_xlabel('')
    ax.set_ylabel('')
    
    # Add the value text on top of the bars for instant readability
    for bar in bars:
        height = bar.get_height()
        if height > 0: # Only label if there is an actual bar
            ax.text(bar.get_x() + bar.get_width()/2., height + 0.02,
                    f'{height:.2f}',
                    ha='center', va='bottom', color='white', 
                    fontweight='bold', fontsize=14)
            
    # Save the plot as a separate image with a transparent background
    plt.tight_layout()
    plt.savefig(filename, transparent=True, dpi=300)
    plt.close()

# --- 1. First Chart Data (Human Distribution) ---
chart1_data = {'No': 0.36, 'Yes': 0.64}
generate_chart(chart1_data, 'slide5_human_dist.png')

# --- 2. Second Chart Data (LLM Stereotype Collapse) ---
chart2_data = {'No': 0.0, 'Yes': 1.00}
generate_chart(chart2_data, 'slide5_llm_dist.png')

print("Charts successfully saved as 'slide5_human_dist.png' and 'slide5_llm_dist.png'")
```
